```python
import networkx as nx
from networks_fenicsx import network_generation, NetworkMesh
```

```python
G = nx.DiGraph([(0, 1), (1, 2)])
G.nodes[0]["pos"] = [0, 0]
G.nodes[1]["pos"] = [0, 0.5]
G.nodes[2]["pos"] = [0, 1]

network_mesh = NetworkMesh(G, N=4)
```

```python
network_mesh.mesh.geometry.x
```

```python
network_mesh.boundaries.values
```

```python
network_mesh.out_marker
```

```python
import pyvista
import dolfinx.plot as plot

cells, types, x = plot.vtk_mesh(network_mesh.mesh)

grid = pyvista.UnstructuredGrid(cells, types, x)

grid.
grid.set_active_scalars("Marker")

plotter = pyvista.Plotter()
plotter.add_mesh(
    grid,
    show_edges=True,
)
plotter.show_axes()
plotter.view_xy()
plotter.show()
```
